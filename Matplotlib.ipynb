# Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
# JupyterLabに対して実行する際に「%」を使う
%matplotlib inline
```

## plt.plot()

```python
x = [1,2,3,4,5]
# y = [1,4,9,16,25]
y = np.array(x) **2
plt.plot(x,y)
```

```python
# plot()にxを入れていないので yの値が 横軸になる
plt.plot(y)
```

```python
# シグモイド関数を描いてみよう
x = np.linspace(-10,10,10)
z = 1 / (1+ np.exp(-x))
plt.plot(x,z)
```

```python
df = pd.DataFrame({'a': [1,2,3,4,5,6],
                   'b':[10,15,40, 100, 200, 400]})
plt.plot(df['a'], df['b'])
```

### 装飾

```python
x = np.linspace(-10, 10,100)
z = 1 / (1 + np.exp(-x))
```

```python
#色を変えれる r=red g=green b=blue, lw=linewidth=太さ, ls =線の種類, marker='o'
plt.plot(x,z, 'r', linewidth=5)
```

```python
plt.plot(x,z,color='r',lw=2, ls='-', marker='o', markersize=10, markerfacecolor='yellow')
```
